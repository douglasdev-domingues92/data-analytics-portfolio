# Análise Exploratória de Turnover em People Analytics

Este notebook apresenta uma análise exploratória sobre fatores associados ao turnover de funcionários, com foco em identificar padrões relevantes para estratégias de retenção.

```python
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.float_format', '{:.2f}'.format)
```

```python
import duckdb

df_raw = pd.read_csv("../data/raw/dataset-ibm.csv")
df = df_raw.copy()

sql = duckdb.connect()
sql.register("employees_raw", df_raw)

sql.execute("""
SELECT
    COUNT(*) AS total_funcionarios,
    SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) AS total_turnover,
    ROUND(100.0 * SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) / COUNT(*), 2) AS turnover_rate
FROM employees_raw
""").df()
```

```python
# Removendo colunas sem potencial valor analítico

colunas_remover = [
    "qtd_funcionarios",
    "matricula_func",
    "maior_18_anos",
    "horas_padrao"
]

df = df.drop(columns=colunas_remover)
df.head()
```

```python
sql.register("employees", df)

sql.execute("""
SELECT
    departamento,
    COUNT(*) AS total_funcionarios,
    SUM(CASE WHEN turnover = 'Yes' THEN 1 ELSE 0 END) AS total_turnover,
    ROUND(
        100.0 * SUM(CASE WHEN turnover = 'Yes' THEN 1 ELSE 0 END) / COUNT(*), 2
        ) AS taxa_turnover
FROM employees
    GROUP BY departamento
    ORDER BY taxa_turnover DESC
""").df()
```

```python
sql.execute("""
SELECT
    horas_extras,
    COUNT(*) AS total_funcionarios,
    SUM(CASE WHEN turnover = 'Yes' THEN 1 ELSE 0 END) AS total_turnover,
    ROUND(
        100.0 * SUM(CASE WHEN turnover = 'Yes' THEN 1 ELSE 0 END) / COUNT(*), 2
        ) AS taxa_turnover
FROM employees
    GROUP BY horas_extras
""").df()
```

```python
import numpy as np

bins = [0, 3000, 5000, 8000, 12000, np.inf]

labels = [
    'Até 3k',
    '3k-5k',
    '5k-8k',
    '8k-12k',
    '12k+'
]

df['faixa_renda'] = pd.cut(
    df['renda_mensal'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
sql.register("employees", df)

sql.execute("""
SELECT
    faixa_renda,
    COUNT(*) AS total_funcionarios,
    SUM(CASE WHEN turnover = 'Yes' THEN 1 ELSE 0 END) AS total_turnover,
    ROUND(
        100.0 * SUM(CASE WHEN turnover = 'Yes' THEN 1 ELSE 0 END) / COUNT(*), 2
    ) AS taxa_turnover
FROM employees
GROUP BY faixa_renda
ORDER BY taxa_turnover DESC
""").df()
```

```python
bins = [-1, 0, 2, 4, np.inf]

labels = [
    '0 anos',
    '1-2',
    '3-4',
    '5+'
]

df['anos_sem_promocao'] = pd.cut(
    df['anos_desde_ultima_promocao'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
```

```python
sql.register("employees", df)

sql.execute("""
SELECT
    anos_sem_promocao,
    COUNT(*) AS total_funcionarios,
    SUM(CASE WHEN turnover = 'Yes' THEN 1 ELSE 0 END) AS total_turnover,
    ROUND(
        100.0 * SUM(CASE WHEN turnover = 'Yes' THEN 1 ELSE 0 END) / COUNT(*), 2
        ) AS taxa_turnover
FROM employees
    GROUP BY anos_sem_promocao
    ORDER BY taxa_turnover DESC
""").df()
```
